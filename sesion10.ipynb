3. ¿Qué son los vectores de soporte?
Aquí está una de las ideas principales de SVM.

Los vectores de soporte son los puntos de entrenamiento que quedan más cerca de la frontera de decisión y que determinan la posición del margen/hiperplano.

Por ejemplo:

🔵 🔵   🔵       |       🔴   🔴 🔴
       ⭐         |         ⭐
🔵 🔵            |            🔴

Los ⭐ serían vectores de soporte.

Una característica importante:

Los puntos que están lejos de la frontera tienen relativamente poca influencia sobre dónde queda la frontera; los vectores de soporte son los ejemplos críticos.

Por eso se llama Support Vector Machine.
