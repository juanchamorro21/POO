Árboles de decisión
Los árboles de decisión son modelos de aprendizaje automático utilizados para resolver problemas de clasificación y regresión. Su estructura está formada por nodos, ramas y hojas. El modelo toma decisiones mediante condiciones sobre las características de los datos hasta llegar a una predicción.

Modelos de caja blanca
Los árboles de decisión son considerados modelos de caja blanca porque su funcionamiento puede interpretarse fácilmente. Es posible observar las características utilizadas, las condiciones de cada división y el camino que sigue cada dato para llegar a una predicción.

Entrenamiento de árboles
Durante el entrenamiento, el algoritmo busca las mejores divisiones de los datos para formar el árbol. En clasificación, las divisiones buscan separar las clases de la manera más adecuada posible. Para determinar la calidad de una división pueden utilizarse medidas como Gini y entropía.

El entrenamiento continúa creando nodos y ramas hasta cumplir determinadas condiciones de parada.

Predicciones
Una vez entrenado, el árbol puede utilizarse para realizar predicciones sobre nuevos datos.

En clasificación, la predicción corresponde a una categoría o clase.

En regresión, la predicción corresponde a un valor numérico.

Probabilidades
Los árboles de clasificación pueden calcular la probabilidad de pertenencia a cada clase. Estas probabilidades se obtienen a partir de la distribución de las clases de las observaciones que llegan a la hoja correspondiente.

CART
CART (Classification and Regression Trees) es un método utilizado para construir árboles de decisión tanto para clasificación como para regresión.

En clasificación, CART utiliza normalmente el índice Gini para seleccionar las divisiones.

En regresión, busca divisiones que reduzcan la variabilidad o el error dentro de los grupos resultantes.

Regularización de árboles
Los árboles pueden crecer demasiado y aprender características específicas de los datos de entrenamiento, provocando sobreajuste (overfitting).

La regularización permite controlar el tamaño y complejidad del árbol. Algunos parámetros importantes son:

max_depth: limita la profundidad máxima.

min_samples_split: establece el mínimo de observaciones necesarias para dividir un nodo.

min_samples_leaf: establece el mínimo de observaciones que puede contener una hoja.

max_leaf_nodes: limita el número de hojas.

ccp_alpha: controla la poda mediante complejidad de costo.

Árboles de regresión
Los árboles de regresión se utilizan cuando la variable objetivo es numérica. El árbol divide los datos en grupos y, en cada hoja, produce un valor de predicción.

Su objetivo durante el entrenamiento es encontrar divisiones que reduzcan el error dentro de los grupos.

Resumen general
Los árboles de decisión son modelos interpretables y versátiles. Pueden utilizarse para clasificación y regresión, permiten obtener predicciones y probabilidades en clasificación y pueden regularizarse para disminuir el sobreajuste. CART es uno de los métodos fundamentales para construir este tipo de árboles.
