🤖 Datos no supervisados

Es una técnica de inteligencia artificial donde el sistema trabaja con datos sin etiquetas o respuestas previamente definidas. Busca encontrar patrones, grupos o comportamientos dentro de los datos.

📊 Agrupamiento y K-Means

El agrupamiento (clustering) consiste en organizar datos similares en grupos. K-Means es un algoritmo que divide los datos en un número determinado de grupos (K), buscando que los elementos de cada grupo sean lo más parecidos posible.

🎯 Segmentación

Consiste en dividir un conjunto de datos en diferentes grupos según características comunes. Por ejemplo, dividir clientes según edad, compras o intereses.

🔵 DBSCAN y formas

DBSCAN es un algoritmo de agrupamiento que identifica grupos basándose en la densidad de los datos. Puede encontrar grupos con formas irregulares y también detectar datos aislados o ruido.

⚠️ Anomalías y novedades

Busca identificar datos que se comportan de manera diferente a lo normal. Es útil para detectar fraudes, errores, comportamientos extraños o situaciones nuevas.

🔔 Mezclas Gaussianas

Las Mezclas Gaussianas (GMM) son un método que representa los datos como una combinación de varias distribuciones estadísticas. Permiten identificar grupos y calcular qué tan probable es que cada dato pertenezca a determinado grupo.

En resumen: todos estos temas forman parte del aprendizaje no supervisado, cuyo objetivo principal es descubrir automáticamente grupos, patrones, formas y anomalías dentro de datos que no tienen etiquetas.
