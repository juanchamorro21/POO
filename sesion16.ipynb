Resumen: Redes Neuronales con Keras
Las redes neuronales artificiales son modelos de aprendizaje automático inspirados en el funcionamiento de las neuronas biológicas. En este tema se estudia cómo funcionan y cómo se pueden implementar utilizando Keras, una biblioteca de alto nivel para construir y entrenar modelos de inteligencia artificial.

1. Biología y redes neuronales
Las redes neuronales artificiales toman como inspiración el cerebro humano. Una neurona biológica recibe señales, las procesa y transmite una respuesta. De forma similar, una neurona artificial recibe diferentes valores de entrada, les asigna pesos, realiza una operación matemática y genera una salida.

El objetivo de las redes neuronales es aprender patrones a partir de datos mediante el ajuste de esos pesos.

2. El Perceptrón
El perceptrón es uno de los modelos más básicos de una neurona artificial. Recibe varias entradas y calcula una combinación de ellas utilizando pesos:

Entradas → pesos → suma → función de activación → salida

Durante el aprendizaje, los pesos se modifican para que el modelo produzca respuestas cada vez más adecuadas.

3. MLP y retropropagación
Un MLP (Multi-Layer Perceptron) es una red neuronal formada por varias capas:

Capa de entrada: recibe los datos.

Capas ocultas: procesan la información y encuentran patrones.

Capa de salida: produce el resultado final.

Para entrenar estas redes se utiliza principalmente la retropropagación (backpropagation). El modelo calcula el error entre su predicción y el resultado esperado y utiliza ese error para ajustar los pesos de las conexiones.

De esta manera, la red aprende progresivamente a realizar una determinada tarea.

4. API Secuencial de Keras
La API Sequential permite crear redes neuronales de manera sencilla, colocando las capas una después de otra.

Es especialmente útil cuando el modelo tiene una estructura lineal:

Entrada → Capa 1 → Capa 2 → Capa 3 → Salida

Es una de las formas más fáciles de comenzar a construir modelos con Keras.

5. API Funcional
La API Functional ofrece mayor flexibilidad que Sequential. Permite crear modelos en los que las capas pueden tener diferentes conexiones.

Es útil cuando se necesitan:

Varias entradas.

Varias salidas.

Conexiones entre diferentes partes de la red.

Arquitecturas más complejas.

Por tanto, permite representar redes que no siguen una estructura estrictamente lineal.

6. API de Subclases
La API de subclases (Subclassing) proporciona todavía más control sobre el modelo. El programador puede definir una clase que represente la red y especificar cómo funcionan sus capas y operaciones.

Esta opción es apropiada para modelos personalizados o arquitecturas muy específicas, aunque requiere mayor conocimiento de Keras y programación.

🧠 Idea principal
El tema muestra una evolución desde los conceptos básicos hasta la implementación:

Neurona biológica → neurona artificial → perceptrón → MLP → retropropagación → construcción de modelos con Keras.

En conclusión, Keras permite crear redes neuronales de diferentes niveles de complejidad. La API Sequential facilita los modelos sencillos, la API Functional permite arquitecturas más flexibles y la API de Subclassing ofrece un mayor nivel de personalización. Todo esto permite desarrollar modelos capaces de aprender patrones a partir de datos y realizar predicciones.
