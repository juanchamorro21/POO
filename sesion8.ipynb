Investigación sobre MNIST

MNIST es uno de los conjuntos de datos más utilizados para aprender y probar algoritmos de inteligencia artificial y aprendizaje automático. Su nombre viene de Modified National Institute of Standards and Technology.

¿Qué contiene?

MNIST contiene imágenes de números escritos a mano, desde el 0 hasta el 9. Cada imagen tiene un tamaño de 28 × 28 píxeles, es decir, 784 píxeles en total.

El conjunto tradicional contiene:

* 60.000 imágenes para entrenamiento.
* 10.000 imágenes para prueba.
* Cada imagen tiene una etiqueta que indica qué número representa.

¿Para qué se utiliza?

Principalmente se utiliza para entrenar y evaluar modelos capaces de reconocer números escritos a mano. Por ejemplo, un modelo puede recibir una imagen y determinar si corresponde al número 3, 7, 8, etc.

Relación con los datos no supervisados

MNIST también puede utilizarse para estudiar aprendizaje no supervisado. Aunque sus imágenes tienen etiquetas, estas pueden ignorarse y utilizarse algoritmos como:

* K-Means: agrupa imágenes que tienen características similares.
* DBSCAN: busca grupos según la densidad de los datos.
* Mezclas Gaussianas: identifica grupos mediante distribuciones estadísticas.
* Detección de anomalías: permite buscar imágenes que se comporten de forma diferente al resto.

Ejemplo sencillo

Si introducimos miles de imágenes de MNIST en K-Means sin decirle qué número representa cada imagen, el algoritmo intentará formar grupos basándose únicamente en las características visuales de los números.

En conclusión: MNIST es un conjunto de datos de imágenes de números escritos a mano que sirve como referencia para estudiar y comparar técnicas de machine learning, clasificación, agrupamiento y reconocimiento de imágenes.
