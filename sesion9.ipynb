MODELO PREDICTIVO PARA ESTIMAR EL PRECIO DE VIVIENDAS EN BOGOTÁ D.C.
1. Introducción
En este proyecto voy a desarrollar un modelo predictivo que permita estimar el precio de viviendas ubicadas en Bogotá D.C., Colombia, utilizando técnicas de aprendizaje automático.

Para realizar el análisis voy a utilizar un conjunto de datos de viviendas de Bogotá que contiene diferentes características relacionadas con las propiedades, como el área construida, número de habitaciones, número de baños, número de garajes, antigüedad y estrato.

El objetivo principal es comparar el rendimiento de dos modelos de Machine Learning: Regresión Lineal y Árbol de Decisión.

La comparación permitirá analizar el comportamiento de los dos modelos para predecir el precio de las viviendas en Bogotá.

Para evaluar los modelos utilizaré diferentes métricas de rendimiento, principalmente el Error Absoluto Medio (MAE), el Error Cuadrático Medio (MSE), la Raíz del Error Cuadrático Medio (RMSE) y el coeficiente de determinación R².

El proceso se desarrollará en Google Colab utilizando Python y las librerías especializadas para análisis de datos y aprendizaje automático.

16. MODELO 1: REGRESIÓN LINEAL
La regresión lineal intenta encontrar una relación matemática entre las características de las viviendas y su precio.

Crear el modelo
# Creo el modelo de regresión lineal.
modelo_lineal = LinearRegression()
Entrenar el modelo
# Entreno la regresión lineal utilizando los datos de entrenamiento.
modelo_lineal.fit(X_train, y_train)
Realizar predicciones
# Utilizo el modelo entrenado para predecir los precios del conjunto de prueba.
predicciones_lineal = modelo_lineal.predict(X_test)
17. Evaluar la regresión lineal
Error Absoluto Medio — MAE
# Calculo el MAE para conocer cuánto se alejan en promedio
# las predicciones del precio real.
mae_lineal = mean_absolute_error(y_test, predicciones_lineal)

print("MAE Regresión Lineal:", mae_lineal)
El MAE se expresa en las mismas unidades del precio.

Como el precio está expresado en pesos colombianos, el MAE también estará expresado en pesos colombianos.

Un MAE menor significa que, en promedio, las predicciones presentan un menor error absoluto.

Error Cuadrático Medio — MSE
# Calculo el MSE para penalizar con mayor fuerza los errores grandes.
mse_lineal = mean_squared_error(y_test, predicciones_lineal)

print("MSE Regresión Lineal:", mse_lineal)
Raíz del Error Cuadrático Medio — RMSE
# Calculo el RMSE para expresar nuevamente el error
# en las mismas unidades monetarias del precio.
rmse_lineal = np.sqrt(mse_lineal)

print("RMSE Regresión Lineal:", rmse_lineal)
Coeficiente de determinación R²
# Calculo R² para conocer qué proporción de la variación
# de los precios logra explicar el modelo.
r2_lineal = r2_score(y_test, predicciones_lineal)

print("R² Regresión Lineal:", r2_lineal)
18. MODELO 2: ÁRBOL DE DECISIÓN
Ahora voy a construir un segundo modelo utilizando un árbol de decisión para regresión.

El árbol intenta dividir los datos en diferentes grupos tomando decisiones basadas en las características de las viviendas.

Crear el modelo
# Creo el árbol de decisión.
modelo_arbol = DecisionTreeRegressor(random_state=42)
Entrenar el árbol
# Entreno el árbol utilizando los mismos datos de entrenamiento.
modelo_arbol.fit(X_train, y_train)
Realizar predicciones
# Utilizo el árbol entrenado para predecir los precios del conjunto de prueba.
predicciones_arbol = modelo_arbol.predict(X_test)
19. Evaluar el árbol de decisión
MAE
# Calculo el MAE del árbol.
mae_arbol = mean_absolute_error(y_test, predicciones_arbol)

print("MAE Árbol de Decisión:", mae_arbol)
MSE
# Calculo el MSE del árbol.
mse_arbol = mean_squared_error(y_test, predicciones_arbol)

print("MSE Árbol de Decisión:", mse_arbol)
RMSE
# Calculo el RMSE del árbol.
rmse_arbol = np.sqrt(mse_arbol)

print("RMSE Árbol de Decisión:", rmse_arbol)
R²
# Calculo R² del árbol.
r2_arbol = r2_score(y_test, predicciones_arbol)

print("R² Árbol de Decisión:", r2_arbol)
